Step 3: preprocessing and DataLoaders.

```python
import torch
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from torch.utils.data import TensorDataset, DataLoader
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.2, stratify=y, random_state=42)
```

```python
X_train, X_test = X_train.copy(), X_test.copy()
```

```python
# fill nulls in ca and thal with the train mode
for col in ["ca", "thal"]:
    mode = X_train[col].mode()[0]
    X_train[col] = X_train[col].fillna(mode)
    X_test[col] = X_test[col].fillna(mode)
```

```python
# one-hot encode categorical columns

cat_cols = ["cp", "restecg", "slope", "thal"]
X_train = pd.get_dummies(X_train, columns=cat_cols, dtype=float)
X_test = pd.get_dummies(X_test, columns=cat_cols, dtype=float)
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)
```

```python
# scale (fit on train only)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

Converting the data into tensors


```python
# to tensors
X_train_t = torch.tensor(X_train, dtype=torch.float32)
X_test_t = torch.tensor(X_test, dtype=torch.float32)
y_train_t = torch.tensor(y_train.values, dtype=torch.float32).unsqueeze(1)
y_test_t = torch.tensor(y_test.values, dtype=torch.float32).unsqueeze(1)
```

```python
# datasets and loaders
train_loader = DataLoader(TensorDataset(X_train_t, y_train_t), batch_size=32, shuffle=True)
test_loader = DataLoader(TensorDataset(X_test_t, y_test_t), batch_size=32)
```

```python
print(X_train_t.shape, X_test_t.shape)
```

```python
print(y_train.value_counts(), y_test.value_counts())
```

```python
from sklearn.model_selection import train_test_split

idx_tr, idx_val = train_test_split(
    range(len(X_train_t)), test_size=0.15,
    stratify=y_train_t.squeeze().numpy(), random_state=42
)

X_val_t, y_val_t = X_train_t[idx_val], y_train_t[idx_val]
train_loader = DataLoader(
    TensorDataset(X_train_t[idx_tr], y_train_t[idx_tr]), batch_size=32, shuffle=True
)
```

```python
import torch.nn as nn


# if gpu fails during the training. cpu will be used instead.

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
class HeartANN(nn.Module):
    def __init__(self, in_features):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_features, 32),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(32, 16),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(16, 1),
        )

    def forward(self, x):
        return self.net(x)  # raw logits, no sigmoid
```

```python
model = HeartANN(X_train_t.shape[1]).to(device)
```

```python
criterion = nn.BCEWithLogitsLoss()
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3, weight_decay=1e-4)
```

```python
print(model)
```

```python
print(sum(p.numel() for p in model.parameters()), "parameters")
```

Training poart

```python
import copy

best_val_loss, wait, best_state = float("inf"), 0, None

for epoch in range(200):
    model.train()
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        criterion(model(xb), yb).backward()
        optimizer.step()

    model.eval()
    with torch.no_grad():
        val_loss = criterion(model(X_val_t.to(device)), y_val_t.to(device)).item()

    if val_loss < best_val_loss:
        best_val_loss, wait = val_loss, 0
        best_state = copy.deepcopy(model.state_dict())
    else:
        wait += 1
        if wait >= 15:
            print("early stopping at epoch", epoch)
            break

    if epoch % 10 == 0:
        print(f"epoch {epoch} | val loss {val_loss:.4f}")

model.load_state_dict(best_state)
```

```python
from sklearn.metrics import (
    confusion_matrix, classification_report, roc_auc_score, accuracy_score
)

model.eval()
with torch.no_grad():
    probs = torch.sigmoid(model(X_test_t.to(device))).cpu().numpy().ravel()

preds = (probs > 0.5).astype(int)
y_true = y_test_t.numpy().ravel().astype(int)

print("accuracy:", accuracy_score(y_true, preds))
print("roc-auc:", roc_auc_score(y_true, probs))
print(confusion_matrix(y_true, preds))
print(classification_report(y_true, preds, target_names=["no disease", "disease"]))
```
